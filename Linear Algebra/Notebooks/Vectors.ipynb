# Vector Additon & Subtraction

```python
def visualise_the_2D_vector_change(v1 : np.array,v2 : np.array, v3 : np.array):
    """
    The point of this function is to show the changes a vector undergoes
    when we manipulate them. Simply, we are visualising the changes as we 
    make them!
    """
    xx, yy = v2.max() + 2, v2.min() - 2 #* -1 + 1

    # plot the vectors
    plt.plot([0, v2[0]],[0, v2[1]], color='blue', label=f'New Vector {v2}')
    plt.plot([0,v3[0]]+v1[0],[0,v3[1]]+v1[1], linestyle = '--',alpha=0.6, label="The Geometric Change")
    plt.plot([0, v1[0]],[0, v1[1]], color='grey', alpha=0.5, label=f'Old Vector {v1}')

    plt.plot([yy,xx], [0,0], color='black', alpha=0.5, linestyle='--')
    plt.plot([0,0], [yy,xx], color='black', alpha=0.5, linestyle='--')

    plt.grid(
        linestyle='--',
        alpha=0.3
    )

    # control the limitsb of the axix
    plt.axis((yy,xx,yy,xx))

    plt.legend()

    plt.title(
        "A Visual Representation of a 2D Vector",
        fontsize=18,
        fontweight='bold',
        color='grey'
    )


v2 = np.array([1,1])
add_vector = np.array([1,2])
v2_new = v2 + add_vector

visualise_the_2D_vector_change(v2,v2_new,add_vector)
```

```python
v2 = np.array([4,1])
v2_add = np.array([-1,-2])
v2_new = v2 + v2_add

visualise_the_2D_vector_change(v2,v2_new,v2_add)
```

```python
v2 = np.array([4,1])
v2_add = np.array([2,-5])
v2_new = v2 + v2_add

# show results
visualise_the_2D_vector_change(v2,v2_new,v2_add)
```

# Vector-scalar multiplication

```python
def vis_the_scaler_impact(v1 : np.array, v2: np.array, l : float):
    """
    The point of this function is to show the changes a vector undergoes
    when we manipulate them. Simply, we are visualising the changes as we 
    make them!
    """
    # Dynamic limits accounting for origin (0,0), v1, and v2
    all_x = [0, v1[0], v2[0]]
    all_y = [0, v1[1], v2[1]]
    
    pad = 2
    x_min, x_max = min(all_x) - pad, max(all_x) + pad
    y_min, y_max = min(all_y) - pad, max(all_y) + pad

    plt.figure(
        figsize=(10,8)
    )
    # plot the vectors
    plt.plot([0, v2[0]],[0, v2[1]], color='red', alpha=0.5, label=f'New Vector {v2}')
    plt.plot([0, v1[0]],[0, v1[1]], color='black', alpha=0.5, linestyle='--',label=f'Old Vector {v1}')

    # Reference axis lines spanning the dynamic limits
    plt.plot([x_min, x_max], [0, 0], color='black', alpha=0.5, linestyle='--')
    plt.plot([0, 0], [y_min, y_max], color='black', alpha=0.5, linestyle='--')

    plt.grid(
        linestyle='--',
        alpha=0.3
    )

    # control the limits of the axis dynamically
    plt.xlim(x_min, x_max)
    plt.ylim(y_min, y_max)

    plt.legend(**legend_params)

    plt.title(
        rf"A Visual Representation of a Scalar Stretching $\lambda={l}$",
        fontsize=18,
        fontweight='bold',
        color='grey'
    )
```

```python
v3 = np.array([1,2])
lam = 0.5
v_new = v3 * lam

vis_the_scaler_impact(v3, v_new, lam)
```

```python
v3 = np.array([1,2])
lam = -1
v_new = v3 * lam

vis_the_scaler_impact(v3, v_new, lam)
```

```python
v3 = np.array([1,2])
lam = 2
v_new = v3 * lam

vis_the_scaler_impact(v3, v_new, lam)
```

# The Geometric Interpretation of Dot Product

```python
v1 = np.array([2, 0, 1])
v2 = np.array([1, 3, 3])

# calculate the distance of each vector
v1_norm = np.linalg.norm(v1)
v2_norm = np.linalg.norm(v2)

print(f"V1 Distance ||a|| = {v1_norm:.2f}")
print(f"V2 Distance ||b|| = {v2_norm:.2f}")

# theta = arccos(aTb / (||a||||b||))
# where ||a|| = sqrt(sum(element^2)) = the vector length
ang = np.arccos(np.dot(v1,v2) / (v1_norm*v2_norm))
deg_ang = np.degrees(ang)

print(f"The Dot Product = {np.dot(v1,v2)}")
print(f"The Geometric Dot Product = {np.linalg.norm(v1)*np.linalg.norm(v2)*np.cos(ang)}")


# show the results
fig = plt.figure(figsize=(10,8));
ax = fig.add_subplot(projection='3d');

# viz the vector
for v in [v1,v2]:
    ax.plot([0,v[0]],[0,v[1]],[0,v[2]]);


# adjust the axis
ax.set_xlabel(r"$X$ axis");
ax.set_ylabel(r"$Y$ axis");
ax.set_zlabel(r"$Z$ axis");
plt.title(f"Angle Between Vectors: $\\theta = {deg_ang:.2f}^\\circ$");
```

# The Cross Product

```python
v1 = np.array([1,2,3])
v2 = np.array([1,1,1])

# calculate the cross prod
cross_prod_v3 = np.cross(v1,v2)

# drop into list to loop and get label
vectors = [
    (v1, r"$\vec{v}_1$"),
    (v2, r"$\vec{v}_2$"),
    (cross_prod_v3, r"$\vec{v}_1 \times \vec{v}_2$ (Cross Product)")
]

print(f"The Cross Product = {cross_prod_v3}")

# show the results
fig = plt.figure(figsize=(10,8));
ax = fig.add_subplot(projection='3d');
n=1
# plot the plane
xx, yy = np.meshgrid(
    np.linspace(-5,5,5),
    np.linspace(-5,5,5)
    )

z1 = (-cross_prod_v3[0]*xx - cross_prod_v3[1]*yy) / cross_prod_v3[2]
ax.plot_surface(xx,yy,z1,alpha=0.25,color='grey')

# viz the vector
for v, label in vectors:
    ax.plot([0,v[0]],[0,v[1]],[0,v[2]],label=f"{label}",linewidth=3);



# adjust the axis
plt.title(f"An Example of the Cross Product", fontsize=18, fontweight='bold',color='grey');
ax.set_xlabel(r"$X$ axis");
ax.set_ylabel(r"$Y$ axis");
ax.set_zlabel(r"$Z$ axis");
ax.grid(
    linestyle='--',
    color='grey',
    alpha=0.5
)
plt.legend(**legend_params);

ax.view_init(azim=30,elev=5)
```

# Cosine Similarity

$$\frac{\mathbf{a}^T \mathbf{b}}{\|\mathbf{a}\| \|\mathbf{b}\|}$$



The Cosine Similarity is exactly the same as the Geometric Interpretation of the the Dot Product!


$$\cos(\theta) ||\mathbf{a}|| ||\mathbf{b}||$$

```python
import pandas as pd
string_a = "Hello, World!"
string_b = "World!"

a_list = string_a.split()
b_list = string_b.split()

# clean the data strings
a_words = [word.replace(",","").replace("!","").replace(".","").lower() for word in a_list]
b_words = [word.replace(",","").replace("!","").replace(".","").lower() for word in b_list]

# join the two together
vocab = sorted(list(set(a_words + b_words)))

# create our dataset
data = [
    {word: a_words.count(word) for word in vocab},
    {word: b_words.count(word) for word in vocab},
]

df = pd.DataFrame(data, index=["string_a", "string_b"])

print(df)

vector_a = df.loc["string_a"].values
vector_b = df.loc["string_b"].values
# lets calculate the cosine similarity between the two strings

#cosine similarity = AB / ||A|| ||B||

#calculate the dot prod
dot_prod = np.dot(vector_a, vector_b)

# calculate the norms
norm_a = np.linalg.norm(vector_a)
norm_b = np.linalg.norm(vector_b)

# now the cosine similarity
cosine_similarity = dot_prod / (norm_a * norm_b)

# convert to an angle
angle = np.arccos(cosine_similarity)

print(f"The Cosine Similarity between the two strings is: {cosine_similarity:.2f}")
print(f"The Angle between the two strings is: {np.degrees(angle):.2f} degrees")
```

```python
# visualise the vectors of bag of words

# normalise the vectors / convert to unit vectors
normal_vector_a = vector_a / np.linalg.norm(vector_a)
normal_vector_b = vector_b / np.linalg.norm(vector_b)

plt.figure(figsize=(10,8))

plt.plot([0, normal_vector_a[0]],[0, normal_vector_a[1]], color='blue', label='Hello World!', linewidth=7)
plt.plot([0, normal_vector_b[0]],[0, normal_vector_b[1]], color='red', label='World!', linewidth=7)

plt.title(
    "A Visual Representation of the Bag of Words Vectors\n"
    r"$\frac{\mathbf{a}^T \mathbf{b}}{\|\mathbf{a}\| \|\mathbf{b}\|}$",
    **title_params
);

plt.legend(
    **legend_params
)

plt.annotate(
    text=(
        f"The Cosine Similarity between\nthe two strings is: {cosine_similarity:.2f}\n"
        f"The Angle between the two\nstrings is: {np.degrees(angle):.2f} degrees"
    ),
    xy=(0, 0),  # Arrow points to (0, 0)
    xytext=(0.25, 0.8),  # Position of the text box
    fontsize=12,
    fontweight='bold',
    ha='center',
    va='center',
    bbox=dict(
        facecolor='grey', 
        alpha=0.5, 
        edgecolor='black', 
        boxstyle='round,pad=0.5'
        ),
    arrowprops=dict(
        arrowstyle='->',
        color='black',
        lw=1.5,
        shrinkB=5  # Adds a small gap so the arrow head doesn't directly clip the origin point
    )
);
```

# Subspace and Span

```python
# Lets use the two vectors from before to visualise the subspace they create


# create a meshgrid of values for lambda and beta
lam_, beta = np.meshgrid(
    np.linspace(-1, 1, 100),  # lambda values
    np.linspace(-3, 3, 100)   # beta values
)

# calculate all possible combinations of the two vectors
x = lam_ * vector_a[0] + beta * vector_b[0]
y = lam_ * vector_a[1] + beta * vector_b[1]


plt.figure(figsize=(10,8))

# fill the spanned region with a light color
plt.fill(
    [x[0,0], x[0,-1], x[-1,-1], x[-1,0]],
    [y[0,0], y[0,-1], y[-1,-1], y[-1,0]],
    color='grey', alpha=0.15, label='Spanned Region'
)

# plot the subspace spanned by the two vectors
plt.plot(x,y, linestyle=':',color='grey', label='Subspace Span', linewidth=1)

# plot the vectors themselves
plt.plot([0, vector_a[0]],[0, vector_a[1]], color='blue', label='Hello World!', linewidth=7)
plt.plot([0, vector_b[0]],[0, vector_b[1]], color='red', label='World!', linewidth=7)
plt.axvline(x=0, color='black', alpha=0.5, linestyle='--')
plt.axhline(y=0, color='black', alpha=0.5, linestyle='--')

# add what spans in span(a,b)
in_text = (
    "The shaded areas is the span of the two vectors.\n"
    "It represents all the points that can be reached\nby linear combinations of the two vectors.\n"
)


plt.annotate(
    in_text,
    xytext=(-0.6, 3.5),  # Position of the text box
    xy=(-0.5, 0.5),
    color='black',
    fontsize=8,
    fontweight='bold',
    ha='center',
    va='center',
    bbox=dict(
        facecolor='lightcoral', 
        alpha=0.5, 
        edgecolor='black', 
        boxstyle='round,pad=0.25',
        pad=0.5
        ),
    arrowprops=dict(
        arrowstyle='->',
        color='black',
        lw=1.5,
        shrinkB=3  # Adds a small gap so the arrow head doesn't directly clip the origin point
    )
)

# add title
plt.title(
    "A Visual Representation of the Subspace Spanned by Two Vectors\n"
    r"$\text{span}(\mathbf{a}, \mathbf{b})$ in $\mathbb{R}^2$",
    **title_params
);

```

## Linear Independence

```python
v1 = np.array([1,2])
v2 = np.array([-1,-2])

# linearly independent vector
v4 = np.array([2,3])
# show the results
fig, ax = plt.subplots(2,1,figsize=(10,8))
ax[0].plot([0, v1[0]],[0, v1[1]], color='blue', label='v1')
ax[0].plot([0, v2[0]],[0, v2[1]], color='red', label='v2', alpha=0.5)

ax[1].plot([0, v1[0]],[0, v1[1]], color='blue', label='v1')
ax[1].plot([0, v4[0]],[0, v4[1]], color='green', label='v4', alpha=0.5)

# adjust visual
for i in range(2):
    ax[i].plot([-4,4], [0,0], color='black', alpha=0.5, linestyle='--')
    ax[i].plot([0,0], [-4,4], color='black', alpha=0.5, linestyle='--')
    ax[i].set_xlim(-4,4)
    ax[i].set_ylim(-4,4)

# add title
plt.suptitle(
    "A Visual Representation of Linearly Dependent & Independent Vectors",
    **title_params
)

sns.despine()
plt.tight_layout()
```
