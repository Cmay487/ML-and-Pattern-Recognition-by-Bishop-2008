# Matrix Types

## Square Matrix

$$A \in \mathbb{R}^{n \times n}$$

```python
# create a matrix
M = np.random.rand(3, 3)

# the result is a Square matrix where M (rows) = N (columns)
print(M)
```

## Rectangular Matrix

$$A \in \mathbb{R}^{m \times n}, \quad m \neq n$$

```python
# create a matrix
M = np.random.rand(3, 4)

# the result is a rectangular matrix where M (rows) != N (columns)
print(M)
```

## Symmetric Matrix

$$A = A^\top \quad \Longleftrightarrow \quad a_{ij} = a_{ji} \;\; \forall\, i, j$$

This is a matrix where the elements are mirrored across the diagonal and the matrix is a Square matrix.

In a way it is like folding a piece of paper and each value is copies on to the oppoiste side!


```python
# create a square matrix
M = np.random.randint(1, 10, (3, 3))

# create the result is a symmetric matrix where M (rows) = N (columns)
M = (M + M.T) / 2

print(M)
```

## Skew-Symmetric Matrix

$$A = -A^\top \quad \Longleftrightarrow \quad a_{ij} = -a_{ji} \;\; \forall\, i, j$$
which implies $a_{ii} = 0$.


This is similar to the Symetric Matrix in its requirements, but the difference is that the mirroring across the diagnal have different signs.

```python
# create a square matrix
M = np.random.randint(1, 10, (3, 3))

# create the result is a symmetric matrix where M (rows) = N (columns)
M = (M - M.T) / 2

print(M)
```

## Identity Matrix
$$I_n = \begin{bmatrix}
1 & 0 & \cdots & 0 \\
0 & 1 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & 1
\end{bmatrix}, \qquad (I_n)_{ij} = \delta_{ij} = \begin{cases} 1 & i = j \\ 0 & i \neq j \end{cases}$$
with the property $AI = IA = A$.

This is a matrix only containing 1 and 0 where 1 lie on the diagonal line. This means that the matrix HAS TO BE SQUARE!

```python
# create the identity matrix where M (rows) = N (columns)
I = np.eye(3)

print(I)
```

## Zero Matrix


$$O_{m \times n} = \begin{bmatrix}
0 & 0 & \cdots & 0 \\
0 & 0 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & 0
\end{bmatrix}, \qquad (O)_{ij} = 0 \;\; \forall\, i, j$$

Does what it says on the tin! A matrix full of zeros

```python
print(np.zeros((3, 3)))
```

## Diagonal Matrix

$$D = \operatorname{diag}(d_1, d_2, \dots, d_n) = \begin{bmatrix}
d_1 & 0 & \cdots & 0 \\
0 & d_2 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & d_n
\end{bmatrix}, \qquad d_{ij} = 0 \;\; \text{for } i \neq j$$

This is similar to the Symetrix and Skew-Symetric matrix, but above and below the diagonal are zero elements and the diagonal elements are non-zero.

```python
# create a diagonal matrix
D = np.diag([1, 2, 3])

D
```

## Upper and Lower Triangular Matrices

### Upper Triangular Matrix
All entries below the diagonal are zero:
$$U = \begin{bmatrix}
u_{11} & u_{12} & \cdots & u_{1n} \\
0 & u_{22} & \cdots & u_{2n} \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & u_{nn}
\end{bmatrix}, \qquad u_{ij} = 0 \;\; \text{for } i > j$$

### Lower Triangular Matrix
All entries above the diagonal are zero:
$$L = \begin{bmatrix}
l_{11} & 0 & \cdots & 0 \\
l_{21} & l_{22} & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
l_{n1} & l_{n2} & \cdots & l_{nn}
\end{bmatrix}, \qquad l_{ij} = 0 \;\; \text{for } i < j$$

These matrices are like Diagonal Matrix but, as the name suggests, zero elements lie on either the top (Upper Triangular Matrix) or bottom (Lower Triangular Matrix) of the diagonal.

```python
# upper triangular matrix
U = np.triu(np.random.rand(3, 3))

U
```

```python
# lower triangular matrix
L = np.tril(np.random.rand(3, 3))

L
```

## Concatenated Matrices

Joining $A \in \mathbb{R}^{m \times n}$ and $B \in \mathbb{R}^{m \times p}$ side by side keeps the number of rows and adds the columns:
$$[A \mid B] \in \mathbb{R}^{m \times (n+p)}$$

$$[A \mid B] = \begin{bmatrix}
a_{11} & \cdots & a_{1n} & b_{11} & \cdots & b_{1p} \\
\vdots & \ddots & \vdots & \vdots & \ddots & \vdots \\
a_{m1} & \cdots & a_{mn} & b_{m1} & \cdots & b_{mp}
\end{bmatrix}$$

This is the method of joining two matrices into one, like a join in SQL!

```python
# create a concatenated matrix
A = np.random.rand(3, 3)
B = np.random.rand(3, 2)

# join
C = np.concatenate([A, B], axis=1)

C
```
